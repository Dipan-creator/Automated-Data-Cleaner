```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.read_csv('messy_customer_data.csv')
```

```python
df = df.dropna(subset=['Customer_ID'])
```

```python
df = df.drop_duplicates()
```

```python
df['Full Name'] = df['Full Name'].str.strip().str.title()
```

```python
df['Full Name'] = df['Full Name'].str.replace('--', '').str.strip()
```

```python
df['Signup_Date'] = pd.to_datetime(df['Signup_Date'], errors='coerce').dt.date
```

```python
df['Phone_Number'] = df['Phone_Number'].astype(str).str.replace(r'[\D]', '', regex=True)
```

```python
df['Phone_Number'] = df['Phone_Number'].str[-10:]
```

```python
df['Phone_Number'] = df['Phone_Number'].replace('', np.nan)
```

```python
df['Purchase_Amount'] = df['Purchase_Amount'].astype(str).str.replace('$', '', regex=False)
```

```python
df['Purchase_Amount'] = pd.to_numeric(df['Purchase_Amount'], errors='coerce')
```

```python
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(0.00)
```

```python
df['Email_Valid'] = df['Email_Address'].str.contains(r'[^@]+@[^@]+\.[^@]+', regex=True)
```

```python
df['Email_Address'] = df['Email_Address'].str.strip().str.lower()
```

```python
df.to_csv('clean_customer_data.csv', index=False)
```

```python
print("Data cleaning complete! Saved as clean_customer_data.csv")
```
